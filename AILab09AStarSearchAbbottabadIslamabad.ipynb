```python
# ===============================================================
# Artificial Intelligence Lab 9
# A* Search Algorithm
# Finding the Optimal Path from Abbottabad to Islamabad
# Objective:
# Implement the A* Search Algorithm to find the optimal route
# between Abbottabad and Islamabad.
#
# A* Evaluation Function:
#
#           f(n) = g(n) + h(n)
#
# where:
# g(n) = Actual cost from the start node to current node
# h(n) = Estimated cost from current node to the goal
#
# ===============================================================

# heapq is used to implement a Priority Queue.
# The node having the smallest f(n) value is expanded first.
import heapq

# ---------------------------------------------------------------
# STEP 1: Create the graph
#
# Each city is represented as a node.
# Roads between cities are represented as edges.
# Numbers represent road distances (km).
# ---------------------------------------------------------------

graph = {

    "Abbottabad": {
        "Havelian": 18
    },

    "Havelian": {
        "Abbottabad": 18,
        "Haripur": 32,
        "Khanpur": 45
    },

    "Haripur": {
        "Havelian": 32,
        "Taxila": 35
    },

    "Khanpur": {
        "Havelian": 45,
        "Taxila": 30
    },

    "Taxila": {
        "Haripur": 35,
        "Khanpur": 30,
        "Islamabad": 28
    },

    "Islamabad": {}
}

# ---------------------------------------------------------------
# STEP 2: Define the heuristic values.
#
# These values estimate the remaining distance to Islamabad.
# The heuristic for the goal node is always zero.
# ---------------------------------------------------------------

heuristic = {

    "Abbottabad": 100,
    "Havelian": 82,
    "Haripur": 55,
    "Khanpur": 50,
    "Taxila": 25,
    "Islamabad": 0
}

# ---------------------------------------------------------------
# STEP 3: Define the A* Search Function
# ---------------------------------------------------------------

def astar(graph, start, goal, heuristic):

    # Priority Queue
    # Stores (f_cost, city)
    open_list = []

    # Insert starting city into the priority queue
    heapq.heappush(open_list, (heuristic[start], start))

    # Stores the parent of every city
    came_from = {}

    # Stores actual travelling cost g(n)
    g_cost = {}

    # Cost of reaching starting city is zero
    g_cost[start] = 0

    # Continue until priority queue becomes empty
    while open_list:

        # Remove node having the smallest f(n)
        current = heapq.heappop(open_list)[1]

        # Goal Test
        if current == goal:
            break

        # Explore all neighbouring cities
        for neighbour, distance in graph[current].items():

            # Compute new travelling cost
            new_cost = g_cost[current] + distance

            # Update if a better path is found
            if neighbour not in g_cost or new_cost < g_cost[neighbour]:

                # Store improved travelling cost
                g_cost[neighbour] = new_cost

                # Compute A* evaluation function
                #
                # f(n)=g(n)+h(n)
                #
                f_cost = new_cost + heuristic[neighbour]

                # Insert into priority queue
                heapq.heappush(open_list, (f_cost, neighbour))

                # Store parent
                came_from[neighbour] = current

    # -----------------------------------------------------------
    # STEP 4: Reconstruct the optimal path
    # -----------------------------------------------------------

    path = []

    city = goal

    while city != start:
        path.append(city)
        city = came_from[city]

    path.append(start)

    path.reverse()

    return path, g_cost[goal]


# ---------------------------------------------------------------
# STEP 5: Define Initial State and Goal State
# ---------------------------------------------------------------

start_city = "Abbottabad"
goal_city = "Islamabad"

# ---------------------------------------------------------------
# STEP 6: Execute A* Search
# ---------------------------------------------------------------

optimal_path, total_distance = astar(
    graph,
    start_city,
    goal_city,
    heuristic
)

# ---------------------------------------------------------------
# STEP 7: Display Results
# ---------------------------------------------------------------

print("=" * 60)
print("Artificial Intelligence Lab 9")
print("A* Search Algorithm")
print("=" * 60)

print("\nInitial State :", start_city)
print("Goal State    :", goal_city)

print("\nOptimal Path")

for city in optimal_path:
    print(city)

print("\nShortest Distance =", total_distance, "km")

print("\nRoute")

print("  →  ".join(optimal_path))

print("\nSearch Completed Successfully.")
```
