```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image

# 1. Load Pre-trained MobileNetV2 Model (Trained on ImageNet's 1000 categories)
print("Loading MobileNetV2 model...")
model = MobileNetV2(weights='imagenet')
```

```python
# Lets give the path of image
img_path = '/content/drive/MyDrive/Trained Model testing/banana.jpg'
img = image.load_img(img_path, target_size=(224, 224))

# 2. Preprocess and predict
x = preprocess_input(np.expand_dims(image.img_to_array(img), axis=0))
preds = model.predict(x, verbose=0)
label = decode_predictions(preds, top=1)[0][0][1].replace('_', ' ').title()
print("The predictedlabel is",label)

# 3. Output prediction and display image
print(f"Predicted Label: {label}")
display(img)
```
